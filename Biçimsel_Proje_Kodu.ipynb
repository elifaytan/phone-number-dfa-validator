```python
import time

class TelefonDFA:
    def __init__(self):
        # Başlangıç durumu
        self.current_state = 'q0'
        # Rakam kümesi
        self.digits = set("0123456789")

    def transition(self, char):
        """
        Format: 0XXX XXX XX XX
        Toplam 14 Karakter (11 rakam + 3 boşluk)
        """
        s = self.current_state
        d = self.digits

        # Hata durumundaysak çık
        if s == 'DEAD': return

        # --- 1. GRUP: 0XXX (Alan Kodu) ---
        if s == 'q0':
            if char == '0': self.current_state = 'q1' # Mutlaka 0 ile başlar
            else: self.current_state = 'DEAD'

        elif s == 'q1': # X
            if char in d: self.current_state = 'q2'
            else: self.current_state = 'DEAD'

        elif s == 'q2': # X
            if char in d: self.current_state = 'q3'
            else: self.current_state = 'DEAD'

        elif s == 'q3': # X (Burada hata veriyordu, şimdi düzeldi)
            if char in d: self.current_state = 'q4'
            else: self.current_state = 'DEAD'

        # --- 1. BOŞLUK ---
        elif s == 'q4':
            if char == ' ': self.current_state = 'q5'
            else: self.current_state = 'DEAD'

        # --- 2. GRUP: XXX ---
        elif s == 'q5':
            if char in d: self.current_state = 'q6'
            else: self.current_state = 'DEAD'

        elif s == 'q6':
            if char in d: self.current_state = 'q7'
            else: self.current_state = 'DEAD'

        elif s == 'q7':
            if char in d: self.current_state = 'q8'
            else: self.current_state = 'DEAD'

        # --- 2. BOŞLUK ---
        elif s == 'q8':
            if char == ' ': self.current_state = 'q9'
            else: self.current_state = 'DEAD'

        # --- 3. GRUP: XX ---
        elif s == 'q9':
            if char in d: self.current_state = 'q10'
            else: self.current_state = 'DEAD'

        elif s == 'q10':
            if char in d: self.current_state = 'q11'
            else: self.current_state = 'DEAD'

        # --- 3. BOŞLUK ---
        elif s == 'q11':
            if char == ' ': self.current_state = 'q12'
            else: self.current_state = 'DEAD'

        # --- 4. GRUP: XX (Bitiş) ---
        elif s == 'q12':
            if char in d: self.current_state = 'q13'
            else: self.current_state = 'DEAD'

        elif s == 'q13':
            if char in d: self.current_state = 'q14' # KABUL DURUMU
            else: self.current_state = 'DEAD'

        else:
            self.current_state = 'DEAD'

    def process_string(self, input_string):
        self.current_state = 'q0'
        print(f"\nİşleniyor: '{input_string}'")

        for char in input_string:
            if self.current_state == 'DEAD': break
            self.transition(char)
            # Adım adım durumu görelim (Debug için)
            # print(f"Girdi: '{char}' -> Durum: {self.current_state}")

        # Sonuç: q14 kabul durumunda olmalı VE uzunluk doğru olmalı
        if self.current_state == 'q14' and len(input_string) == 14:
            return True
        else:
            return False

# --- Test ---
if __name__ == "__main__":
    dfa = TelefonDFA()
    while True:
        print("\n--- Telefon Doğrulama (Format: 0XXX XXX XX XX) ---")
        user_input = input("Numarayı girin (Çıkış 'q'): ")
        if user_input.lower() == 'q': break

        if dfa.process_string(user_input):
            print(f">>> SONUÇ: '{user_input}' GEÇERLİ ✅")
        else:
            print(f">>> SONUÇ: '{user_input}' GEÇERSİZ ❌")
```
